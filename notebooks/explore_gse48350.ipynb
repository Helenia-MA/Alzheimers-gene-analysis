## Building our first metadata table now

```python
import gzip

sample_ids = expression.columns[1:]

with gzip.open("../data/GSE48350_series_matrix.txt.gz", "rt") as f:
    lines = f.readlines()

sample_title_line = lines[35]
titles = sample_title_line.strip().split("\t")[1:]

print("no of titles:", len(titles))
print()

for title in titles[:10]:
    print(title)
```

```python
metadata = pd.DataFrame({
    "sample_ids": sample_ids,
    "titles": titles
})
metadata.head()
```

```python
# adding disease status
metadata["disease"] = metadata["titles"].apply(
    lambda x: "AD" if "AD" in x else "control"
)
metadata["disease"].value_counts()
```
